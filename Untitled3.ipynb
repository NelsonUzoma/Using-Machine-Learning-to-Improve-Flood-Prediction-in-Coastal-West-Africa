Step 2: Select Features and Target

The target variable is median_house_value, which is the value we want the model to predict. I selected median_income and housing_median_age as features because the median income of an area ( for example, let's say in the Chicago Loop is 20, 25 or 30 dollars per hour) and housing age ( for example, let's say the buildings are 5, 10 or 15 years old) may have the strongest relationship with house values.

```python
X = df[['median_income', 'housing_median_age']]
Y = df['median_house_value']
```

```python
X.head()
```

```python
Y.head()
```

Step 3: Split the Dataset

The dataset is been split into 70%  training and 30% testing data. The training data is used to train the model whilst the testing data is reserved to test the model after training to see how well it performs on unseen data.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.30, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
Y_train.shape
```

```python
Y_test.shape
```

Step 4: Train a Linear Regression Model

We use Linear Regression to learn the relationship between the selected features and the target variable using the training data.

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, Y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

Step 5: Make Predictions and Evaluate the Model

The trained model is used to predict house values for the test dataset. The predictions are then compared with the actual house values to evaluate the model's performance.

```python
Y_pred = model.predict(X_test)
```

```python
Y_pred[:10]
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mae = mean_absolute_error(Y_test, Y_pred)
mae
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mse = mean_squared_error(Y_test, Y_pred)
mse
```

```python
from sklearn.metrics import r2_score
```

```python
r2 = r2_score(Y_test, Y_pred)
r2
```

```python
print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error(MSE):", mse)
print("R² Score:", r2)
```

Model Evaluation:

The Linear Regression model achieved a Mean Absolute Error (MAE) of 60,754.67, meaning that the model's predictions differ from the actual median house values by approximately $60,755 on average. The Mean Squared Error (MSE) was 6,669,658,759.17. MSE gives greater weight to larger errors because the errors are squared. The R² score was 0.5171, meaning that the model explains approximately 51.7% of the variation in median house values using the selected features. Overall, the model shows moderate predictive performance, but there is still substantial variation that is not explained by these two features.
